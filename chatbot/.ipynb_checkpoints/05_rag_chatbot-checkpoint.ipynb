# 05 — RAG Auction Chatbot
Retrieval-Augmented Generation chatbot for the Nexus auction platform.

**Architecture:**
1. Knowledge base built from real eBay auction data (Cartier, Xbox, Palm)
2. SBERT retrieves relevant context for each user question
3. ChromaDB stores and searches the knowledge base
4. Ollama (llama3.2) generates natural language responses
5. Connects to your /recommend API for personalized suggestions

**Prerequisites:**
- `ollama pull llama3.2` already done
- `pip install ollama chromadb` already done
- Notebooks 01-04 already run
- auction.csv copied to chatbot/datasets/

```python
import pandas as pd
import numpy as np
import chromadb
import ollama
import os, json, re, requests
from sentence_transformers import SentenceTransformer

DATA_DIR    = '../data/processed'
DATASET_DIR = '../chatbot/datasets'
CHROMA_DIR  = '../chatbot/chroma_db'

os.makedirs(DATASET_DIR, exist_ok=True)
os.makedirs(CHROMA_DIR,  exist_ok=True)

print('Imports OK')
```

## 1. Load Auction Datasets

```python
auction_df = pd.read_csv(f'{DATASET_DIR}/auction.csv')
print(f'Loaded: {len(auction_df):,} rows')
print(f'Items: {auction_df["item"].unique()}')
display(auction_df.head(3))
```

```python
# Key stats the chatbot will use as facts
stats = auction_df.groupby('item').agg(
    total_bids      = ('bid', 'count'),
    avg_final_price = ('price', 'mean'),
    min_price       = ('price', 'min'),
    max_price       = ('price', 'max'),
    avg_open_bid    = ('openbid', 'mean'),
).round(2).reset_index()

display(stats)
```

## 2. Build Knowledge Base Documents

```python
documents = []

# A. Data-derived price facts
for _, row in stats.iterrows():
    item = row['item']
    doc  = (
        f"Auction statistics for {item}: "
        f"Average closing price is £{row['avg_final_price']:.0f}. "
        f"Prices range from £{row['min_price']:.0f} to £{row['max_price']:.0f}. "
        f"Average opening bid is £{row['avg_open_bid']:.0f}. "
        f"Total {row['total_bids']:,} bids recorded."
    )
    documents.append({'id': f'stats_{item.replace(" ","_")}', 'text': doc, 'category': 'statistics'})

# B. Bidding pattern facts
for item in auction_df['item'].unique():
    item_df = auction_df[auction_df['item'] == item]
    early   = (item_df['bidtime'] < 1).sum()
    total   = len(item_df)
    late    = (item_df['bidtime'] > (item_df['bidtime'].max() - 0.5)).sum()
    doc = (
        f"Bidding patterns for {item}: "
        f"{early/total*100:.0f}% of bids in first 24 hours. "
        f"{late/total*100:.0f}% of bids in final 12 hours (sniping). "
        f"Most competitive bidding happens near auction end."
    )
    documents.append({'id': f'patterns_{item.replace(" ","_")}', 'text': doc, 'category': 'patterns'})

# C. Auction type comparison
type_stats = auction_df.groupby('auction_type')['price'].mean().round(0)
doc = (
    'Auction duration comparison: '
    + '. '.join([f'{t} average £{p:.0f}' for t, p in type_stats.items()])
    + '. Longer auctions attract more bidders and higher prices.'
)
documents.append({'id': 'auction_types', 'text': doc, 'category': 'auction_types'})

# D. FAQ documents
faqs = [
    ('faq_bidding',   'faq', 'How to bid: Click Place Bid on any item and enter your maximum amount. The system autobids on your behalf up to your maximum. Highest bidder wins when auction ends.'),
    ('faq_reserve',   'faq', 'Reserve price: The minimum price the seller accepts. If bidding does not reach it the item is unsold and returned. You will see Reserve Met when it is reached.'),
    ('faq_premium',   'faq', "Buyer's premium: Nexus charges 15% on top of the hammer price. A £100 win costs £115 total. This covers auction services authentication and secure payment."),
    ('faq_autobid',   'faq', 'Autobid proxy bidding: Enter your maximum and the system bids the minimum needed to keep you winning up to your limit. Your maximum is never shown to other bidders.'),
    ('faq_winning',   'faq', 'Winning an auction: You get an email confirmation. Payment required within 48 hours. We accept bank transfer credit card and PayPal. Items ship within 3 business days.'),
    ('faq_sniping',   'faq', 'Sniping last minute bidding: Legal strategy of bidding in final seconds to avoid driving up price. Counter it by using autobid with your true maximum.'),
    ('faq_condition', 'faq', 'Item conditions: Mint never used. Excellent minimal wear. Good normal wear. Fair visible wear. Items over £500 come with certificate of authenticity.'),
    ('faq_selling',   'faq', 'Selling on Nexus: Free appraisals above £100. 12% seller commission. We handle photography description and promotion. Payment within 10 business days.'),
    ('faq_categories','faq', 'Categories: Antiques Jewellery Art Ceramics Furniture Watches Books Coins Toys Musical Instruments. New lots added daily.'),
    ('faq_rating',    'faq', 'Bidder ratings: Score based on completed transactions. Higher score means more reliable buyer. Average active buyer rating is 50 to 180.'),
]

for fid, fcat, ftext in faqs:
    documents.append({'id': fid, 'text': ftext, 'category': fcat})

# E. Item catalog chunks
catalog = pd.read_csv(f'{DATA_DIR}/item_catalog.csv')
for cat, group in catalog.groupby(catalog['item_title'].str.split().str[0]):
    if len(group) < 3:
        continue
    titles = group['item_title'].tolist()[:5]
    avg_p  = group['avg_price'].mean()
    doc    = (
        f"Items starting with '{cat}': "
        + ', '.join(titles)
        + f'. Average price £{avg_p:.2f}. Total {len(group)} items.'
    )
    documents.append({'id': f'catalog_{cat[:20]}', 'text': doc, 'category': 'catalog'})

print(f'Total documents: {len(documents)}')
print('By category:', pd.Series([d["category"] for d in documents]).value_counts().to_dict())
```

## 3. Build ChromaDB Vector Store

```python
print('Loading SBERT...')
sbert = SentenceTransformer('all-MiniLM-L6-v2')

chroma_client = chromadb.PersistentClient(path=CHROMA_DIR)

try:
    chroma_client.delete_collection('nexus_knowledge')
except:
    pass

collection = chroma_client.create_collection('nexus_knowledge')

texts      = [d['text']     for d in documents]
ids        = [d['id']       for d in documents]
categories = [d['category'] for d in documents]

print(f'Embedding {len(texts)} documents...')
embeddings = sbert.encode(texts, batch_size=32, show_progress_bar=True).tolist()

collection.add(
    ids        = ids,
    documents  = texts,
    embeddings = embeddings,
    metadatas  = [{'category': c} for c in categories]
)

print(f'Knowledge base: {collection.count()} documents indexed')
```

## 4. Intent Detection + RAG Pipeline

```python
INTENT_PATTERNS = {
    'recommend':        [r'recommend', r'suggest', r'what should i', r'show me', r'find me', r'looking for'],
    'price_query':      [r'how much', r'price', r'cost', r'worth', r'value', r'sell for'],
    'how_to_bid':       [r'how.*bid', r'how.*auction', r'how.*buy', r'place.*bid', r'how.*work'],
    'bidding_strategy': [r'strategy', r'when.*bid', r'snip', r'last minute', r'autobid', r'best time'],
    'greeting':         [r'^hi', r'^hello', r'^hey'],
}

def detect_intent(text):
    text_lower = text.lower()
    for intent, patterns in INTENT_PATTERNS.items():
        for p in patterns:
            if re.search(p, text_lower):
                return intent
    return 'general'


def retrieve(query, n=3):
    emb     = sbert.encode([query]).tolist()
    results = collection.query(query_embeddings=emb, n_results=n)
    return results['documents'][0]


def call_api(endpoint, params={}):
    try:
        r = requests.get(f'http://localhost:8000/{endpoint}', params=params, timeout=5)
        return r.json() if r.ok else {}
    except:
        return {}


SYSTEM_PROMPT = """You are ARIA, the AI assistant for Nexus Auctions, a premium online auction house.
Help users find items, understand bidding, and make smart decisions.
Answer using ONLY the provided context. Be warm, concise, and use British English with £ for prices.
If unsure, say so honestly. Keep responses to 2-4 sentences."""


def chat(user_message, user_id='GUEST', history=None):
    intent        = detect_intent(user_message)
    context_parts = []

    # Retrieve knowledge
    chunks = retrieve(user_message, n=3)
    if chunks:
        context_parts.append('KNOWLEDGE:\n' + '\n'.join(chunks))

    # Add API context for recommendations
    if intent == 'recommend':
        data = call_api('recommend', {'user_id': user_id, 'top_n': 3})
        items = data.get('recommendations', [])
        if not items:
            items = call_api('popular', {'top_n': 3}).get('items', [])
        if items:
            lines = [f"{i+1}. {it.get('item_title','').title()} — £{it.get('avg_price',0):.2f}"
                     for i, it in enumerate(items[:3])]
            context_parts.append('RECOMMENDED ITEMS:\n' + '\n'.join(lines))

    context  = '\n\n'.join(context_parts)
    messages = [{'role': 'system', 'content': SYSTEM_PROMPT}]

    if history:
        messages.extend(history[-6:])

    user_content = f'Context:\n{context}\n\nQuestion: {user_message}' if context else user_message
    messages.append({'role': 'user', 'content': user_content})

    try:
        response = ollama.chat(
            model=OLLAMA_MODEL,
            messages=messages,
            options={'temperature': 0.7, 'num_predict': 200}
        )
        return response['message']['content'], intent
    except Exception as e:
        return f'Sorry, I cannot connect to the AI model. Please ensure Ollama is running. ({e})', 'error'


OLLAMA_MODEL = 'llama3.2'
print('Pipeline ready.')
```

## 5. Test the Chatbot

```python
tests = [
    'Hello! What can you help me with?',
    'How much do Cartier watches sell for?',
    'What is a reserve price?',
    'When is the best time to place a bid?',
    'Can you recommend something for me?',
    "How does the buyer's premium work?",
]

history = []
for q in tests:
    print(f'You: {q}')
    answer, intent = chat(q, user_id='TEST_USER', history=history)
    print(f'ARIA [{intent}]: {answer}')
    print()
    history.append({'role': 'user',      'content': q})
    history.append({'role': 'assistant', 'content': answer})
```

## 6. Interactive Chat

```python
print('ARIA — Nexus Auction Assistant (type quit to exit)')
print('-' * 50)

history = []
while True:
    user_input = input('You: ').strip()
    if not user_input: continue
    if user_input.lower() in ['quit', 'exit', 'bye']:
        print('ARIA: Thank you for visiting Nexus Auctions!')
        break
    response, intent = chat(user_input, history=history)
    print(f'ARIA: {response}\n')
    history.append({'role': 'user',      'content': user_input})
    history.append({'role': 'assistant', 'content': response})
```

## 7. Save Config for API

```python
config = {
    'chroma_dir':    CHROMA_DIR,
    'ollama_model':  OLLAMA_MODEL,
    'system_prompt': SYSTEM_PROMPT,
    'n_results':     3,
}

os.makedirs('../chatbot', exist_ok=True)
with open('../chatbot/chatbot_config.json', 'w') as f:
    json.dump(config, f, indent=2)

print('Saved chatbot_config.json')
print('Chatbot build complete!')
```
